Browse per-session behavioral summaries and diagnostics. The table below loads `data/BG_046_sessions_manifest.csv` and displays clickable thumbnails for psychometric and RT histograms. Click a thumbnail to open the full-size PNG in the notebook output or your browser.

# Learning browser

Interactive notebook to explore session manifest and learning curves for a subject.
Run the first cell to load the manifest produced by `scripts/build_manifest_and_behavior_summary.py`.

```python
import pandas as pd
from IPython.display import HTML, display
import os

manifest_path = os.path.join('..', 'data', 'BG_046_sessions_manifest.csv') if not os.path.exists('data/BG_046_sessions_manifest.csv') else 'data/BG_046_sessions_manifest.csv'

try:
    df = pd.read_csv(manifest_path)
except Exception as e:
    display(HTML(f"<b>Error loading manifest:</b> {e}"))
    df = pd.DataFrame()

if not df.empty:
    def make_thumb(path, alt=''):
        if not path or (not isinstance(path, str)):
            return ''
        p = path.replace('\\', '/')
        if os.path.exists(p):
            return f"<a href='{p}' target='_blank'><img src='{p}' style='height:120px; border:1px solid #ccc; margin:4px;'></a>"
        else:
            return f"<span style='color:#999'>{alt}</span>"

    rows = []
    for _, r in df.iterrows():
        psych = make_thumb(r.get('psych_png', ''), 'no-psych')
        rth = make_thumb(r.get('rthist_png', ''), 'no-rth')
        rows.append(f"<tr><td>{r['session_name']}</td><td>{int(r['n_trials'])}</td><td>{r['n_hits']}</td><td>{r['n_fa']}</td><td>{psych}</td><td>{rth}</td></tr>")

    table = """
    <table style='border-collapse:collapse'>
      <thead><tr><th>Session</th><th>Trials</th><th>Hits</th><th>FAs</th><th>Psychometric</th><th>RT hist</th></tr></thead>
      <tbody>
    """ + "\n".join(rows) + "</tbody></table>"
    display(HTML(table))
else:
    display(HTML('<p>No manifest available.</p>'))
```
